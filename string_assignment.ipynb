```python
##1Characters

#Example

char1 = "I"
char2 = "8"
char3 = "$"
print(char1,char2,char3)
```

```python
##2Strings

#Example

MyFavouriteAnimal = "German Shepherd"
FactAboutIt = "German Shepherds are brave!!!"
print(MyFavouriteAnimal)
print(FactAboutIt)
```

```python
##3Concatenation

#Example
#Sticking words together like Lego Blocks

first_name = "German"
last_name = "Shepherd"
full_name = first_name+ " " +last_name #Joining Strings Together
print(full_name)
```

```python
##4Repetition

#Example
#Repeating a string

flower_name = "Jasmine"
girl_name = "Jennie"
print(flower_name *5)     #Repeating a String 5 times
print(girl_name *3)       #Repeating a String 3 times  
```

```python
##5Indexing

#Example
#In Pyhthon counting starts from 0

Language = "PythonRocks"
print(Language [0])       #Picking the letter P by it's spot 
print(Language [3])       #Picking the letter h by it's spot
print(Language [-1])      #Picking the letter s by it's spot
```

```python
##6Slicing

#Example
#Taking a slice of the word like cutting a cake!

Japan = "Tokyo is the capital city of Japan"
print(Japan [0:5])   #Taking a slice of word from it's string
print(Japan [28:])   #Taking a slice of word from it's sting
```
